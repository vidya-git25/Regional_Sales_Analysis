```python
monthly_sales = (
    df.groupby(pd.Grouper(key='order_date', freq='MS'))['line_total']
      .sum()
      .sort_index()
)
```

```python
print(monthly_sales.head())
```

```python
plt.figure(figsize=(15,6))
plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    color='olive',
    marker='o'
)
import matplotlib.dates as mdates
# Format X-axis labels as Jan 2014, Feb 2014, etc.
ax = plt.gca()
ax.xaxis.set_major_locator(mdates.MonthLocator(interval=3))
ax.xaxis.set_major_formatter(mdates.DateFormatter('%b %Y'))

plt.title('Monthly Sales Trend')
plt.xlabel('Month and Year')
plt.ylabel('Total Sales')
#plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

##### Observations:
###### In the above plot,The highest total sales from jan 2014 to apr 2018 ,is in march2016 and may 2017 followed by Nov 2014.
###### And the lowest total sales is in feb 2017,feb 2015,apr 2016

### Customer Segmentation:Revenue vs Profit Marging

```python
customer_data=df.groupby('customer_names').agg(revenue=('line_total','sum'),
                                               profit=('profit','sum'),
                                               profit_margin_pct=('profit_margin_pct','mean'))
```

```python
customer_data
```

```python
# Calculate revenue, profit, and profit margin for each customer
customer_data = (
    df.groupby('customer_names')
      .agg(
          revenue=('line_total', 'sum'),
          profit=('profit', 'sum'),
          profit_margin_pct=('profit_margin_pct', 'mean')
      )
      .reset_index()
)

# Create scatter plot
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=customer_data,
    x='revenue',
    y='profit_margin_pct',
    hue='profit',
    palette='viridis',
    size='revenue',
    sizes=(40, 250),
    alpha=0.8
)

plt.title('Customer Segmentation: Revenue vs Profit Margin', fontsize=16)
plt.xlabel('Total Revenue', fontsize=13)
plt.ylabel('Average Profit Margin (%)', fontsize=13)

plt.tight_layout()
plt.show()
```
